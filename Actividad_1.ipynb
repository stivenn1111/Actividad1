```python
from typing import List

class Item:
    def __init__(self, nombre: str, precio: float):
        if precio < 0:
            raise ValueError("el precio no puede ser negativo")

        self.__nombre = nombre
        self.__precio = precio

    def get_nombre(self):
        return self.__nombre

    def get_precio(self):
        return self.__precio


class Factura:
    def __init__(self):
        self.__items = []

    def agregar_item(self, item: Item):
        if not isinstance(item, Item):
            raise TypeError("solo se pueden agregar objetos tipo item")
        self.__items.append(item)

    def calcular_total(self):
        if not self.__items:
            raise ValueError("error: no se puede calcular el total si la factura esta vacia")
        return self.__sumar_mitad(0, len(self.__items) - 1)

    def __sumar_mitad(self, inicio: int, fin: int):
        if inicio == fin:
            return self.__items[inicio].get_precio()

        mitad = (inicio + fin) // 2

        total_izq = self.__sumar_mitad(inicio, mitad)
        total_der = self.__sumar_mitad(mitad + 1, fin)

        return total_izq + total_der

```

```python
# prueba 1: encapsulamiento
print("prueba de encapsulamiento")
producto1 = Item("portatil gamer", 3500000.0)
print("producto:", producto1.get_nombre(), "| precio:", producto1.get_precio())

try:
    print(producto1.__precio)
except AttributeError:
    print("no se puede acceder al atributo __precio directamente porque es privado\n")

# prueba 2: calculo recursivo con 100000 items
print("prueba con 100000 items")
factura_masiva = Factura()


for i in range(100000):
    factura_masiva.agregar_item(Item(f"item_{i}", 10.0))

total = factura_masiva.calcular_total()
print("total calculado:", total)
print("funciono bien sin errores de pila\n")

# prueba 3: caso borde
print("prueba caso borde con lista vacia")
factura_vacia = Factura()

try:
    factura_vacia.calcular_total()
except ValueError as e:
    print("error capturado correctamente:", e)
```
